### Purpose & Academic Defense
This notebook explores the official **AI4I 2020 Predictive Maintenance Dataset** from the UCI Machine Learning Repository.

#### Key Methodological Principles:
1. **Strict Target Leakage Prevention:** Identification columns (`UDI`, `Product ID`) and failure-mode cause flags (`TWF`, `HDF`, `PWF`, `OSF`, `RNF`) are strictly audited and excluded from feature set `X`.
2. **Operational Meaning:** The dataset represents row-level operating snapshots. It does **not** provide ordered continuous trajectories per machine, and row-level classification is **never** presented as "failure in N minutes/hours/cycles".
3. **Extreme Class Imbalance:** Normal operating conditions account for ~96.6% of records, with failures at ~3.39%. Evaluation must prioritize Recall, Precision, PR-AUC / Average Precision, and F1-score over raw accuracy.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Set paths
DATA_PATH = Path("../data/raw/ai4i2020.csv")
FIG_DIR = Path("../reports/figures")
FIG_DIR.mkdir(parents=True, exist_ok=True)

print("Loading dataset...")
df = pd.read_csv(DATA_PATH)
print(f"Shape: {df.shape[0]} rows, {df.shape[1]} columns")
df.head()
```

## 1. Feature Audit & Leakage Inspection

```python
APPROVED_FEATURES = [
    "Type",
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]"
]
EXCLUDED_COLUMNS = ["UDI", "Product ID", "TWF", "HDF", "PWF", "OSF", "RNF", "Machine failure"]

print("Approved Candidate Features:", APPROVED_FEATURES)
print("Strictly Excluded Columns:", EXCLUDED_COLUMNS)

# Missing values and data types
df.info()
```

## 2. Target Variable & Failure Mode Distributions

```python
target_counts = df["Machine failure"].value_counts()
target_pct = df["Machine failure"].value_counts(normalize=True) * 100
imbalance_summary = pd.DataFrame({"Count": target_counts, "Percentage (%)": target_pct.round(2)})
print("--- Class Distribution ---")
print(imbalance_summary)

# Failure modes breakdown
failure_modes = ["TWF", "HDF", "PWF", "OSF", "RNF"]
print("\n--- Failure Modes Breakdown ---")
print(df[failure_modes].sum())
```

## 3. Physical Operating Variable Distributions

```python
numeric_features = [col for col in APPROVED_FEATURES if col != "Type"]
df[numeric_features].describe().round(2)
```
